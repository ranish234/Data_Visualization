```python


import seaborn as sns
import matplotlib.pyplot as plt

# Load a built-in dataset
tips = sns.load_dataset("tips")

# 1. Standard filled histogram (Default)
sns.histplot(data=tips, x="total_bill",fill=False)
plt.show()

```

```python
# 2. Unfilled step-histogram
sns.histplot(data=tips, x="total_bill", element="step", fill=True)
plt.show()
```

```python

temp_df = tips.pivot_table(
    index="day", columns="sex", values="total_bill", aggfunc="mean"
)
```

```python
plt.figure(figsize =(2,2))
sns.heatmap(temp_df)
```

```python
penguins = sns.load_dataset('penguins')
```

```python
correlation_matrix = penguins.select_dtypes(include='number').corr()
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Penguin Features Correlation Matrix on Measuremenets')
plt.show()
```
