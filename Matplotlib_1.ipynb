### Coloured Scatterplot:


```python
from sklearn.datasets import load_iris
import pandas as pd

# Load dataset
iris = load_iris(as_frame=True)
df = iris.frame

df.target.value_counts()
```

```python
df.head()
```

```python
df.columns
```

```python
# coloured scatter plot :
plt.figure(figsize = (3,4))

plt.scatter(df['sepal length (cm)'],df['petal length (cm)'],c = df['target'],cmap="jet")
plt.colorbar()
plt.title("Sepal Length V/S Petal Length")
plt.axhline(4,color = "red")
plt.axvline(6.5, color ="yellow")
plt.xlabel("Sepal Length",fontdict ={'color': 'green'})
plt.ylabel("Petal Legth",fontdict ={'color': 'red'})
```

### Subplots:

```python
df.columns
```

```python
fig , ax = plt.subplots(nrows = 2, ncols = 2 ,sharex= True , figsize = (10,6))
ax[0,0].scatter(df['sepal length (cm)'],df['petal length (cm)'] , c = df['target'],cmap = "jet", marker ="+")
ax[0,1].scatter(df['petal length (cm)'],df['petal width (cm)'], c = df['target'],cmap ="jet",marker ="*")
ax[1,0].hist(df['petal length (cm)'])
ax[1,1].hist(df['petal width (cm)'])
ax[0,0].set_title("Iris Flower")
ax[0,1].set_ylabel("petal width")
```

```python
# 3D scatter plot :
fig = plt.figure(figsize = (6,8))
ax = plt.subplot(projection = '3d')
ax.scatter(df['sepal length (cm)'],df['petal length (cm)'],df['target'] ,c =df['target'] ,cmap='jet', alpha= 1)
ax.plot3D(df['sepal length (cm)'],df['petal length (cm)'],df['target'] )
ax.set_title("Iris Flower")
ax.set_xlabel("sepal length")
ax.set_ylabel("petal length")
ax.set_zlabel("petal width")
```

```python
# 3D line plot :
ax = plt.subplot(projection = '3d')
ax.plot3D(df['sepal length (cm)'],df['petal length (cm)'],df['target'] )
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx ,yy = np.meshgrid(x,y)
z = np.cos(xx)+np.sin(yy)
```

```python
# 3d surface plot :
fig = plt.figure(figsize = (7,7))
ax = plt.subplot(projection = "3d")
ax.plot_surface(xx, yy,z,cmap = "jet")
```

```python
# contour plots 
ax = plt.subplot(projection = "3d")
ax.contour(xx , yy,z , cmap = "jet")
```

```python
# Heatmap :
# 1. Create a 2D array
data = np.random.rand(5, 5)

# 2. Initialize plot
fig, ax = plt.subplots(figsize=(7, 5))
cax = ax.imshow(data, cmap='coolwarm', interpolation='nearest')

# 3. Add a colorbar legend
fig.colorbar(cax)

# 4. Set ticks and labels manually
labels = ['A', 'B', 'C', 'D', 'E']
ax.set_xticks(np.arange(len(labels)))
ax.set_yticks(np.arange(len(labels)))
ax.set_xticklabels(labels)
ax.set_yticklabels(labels)

plt.title("Matplotlib Heatmap")
```
